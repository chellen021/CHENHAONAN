## ③ Gemini API 連線測試

執行下方程式單元，以你的 API 金鑰呼叫 Gemini 模型，確認金鑰有效與網路連線正常。預期輸出：「Gemini API 連線測試：OK」＋模型回應。

**失敗處理**：
- 顯示「未偵測到 GEMINI_API_KEY」→ 依上方「設定 Gemini API 金鑰」單元完成設定後，重新執行本單元
- 顯示連線失敗 → 可能原因：金鑰無效／免費層額度用完／網路受限——可改用工具對照表之替代工具（見 Lab_Ch01 指引）

```python
try:
    from google.colab import userdata
    api_key = userdata.get("GEMINI_API_KEY")
except ImportError:
    api_key = None

if not api_key:
    print("⚠️ 未偵測到 GEMINI_API_KEY——請依上方「設定 Gemini API 金鑰」步驟完成後重試。")
    print("（其餘環境驗證不受影響，可先繼續）")
else:
    try:
        import google.generativeai as genai
        genai.configure(api_key=api_key)
        model = genai.GenerativeModel("gemini-3-flash")
        resp = model.generate_content("請以一句話回應：連線測試成功。")
        print("Gemini API 連線測試：OK ✅")
        print("模型回應：", resp.text.strip())
    except Exception as e:
        print(f"⚠️ Gemini API 連線失敗：{type(e).__name__}: {e}")
        print("可能原因：金鑰無效／免費層額度用完／網路受限（可改用工具對照表之替代工具）")
```
